        Through this project, our group seeks to answer the following: Amidst all the differences among bats, what allows them to be placed under an umbrella term? Are the differences across bats insufficient to allow them to be categorized as another organism? In an attempt to explore these questions, we will be comparing echolocation, a key characteristic that a majority of bats hold. Specifically, we seek to look at the frequency of echolocation across a handful of bat species in order to determine whether or not there is a significant enough difference to categorize certain bats as “bat outliers”.
        
        We will be using the Bat Call Library: ChiroVox for our dataset. This library includes audio recordings of real bat echolocations from species across the world, several of which we will be using in our project dataset. We will analyze the bats’ call frequency structures and call durations to determine if there are any differences in bat calls across the species featured in our project. This dataset also provides metadata recordings, which contain specific names, the recording method, the recording device, and the sampling rate associated with the audio recording. This dataset helps to support our goal of identifying the species-specific signatures, as well as to better understand how bats communicate among each other. 
        
        A signal processing technique we will be using is low-pass filtering. Echolocation calls are high-frequency signals, with the recordings containing background noises of even higher frequencies. As a result, we will need to utilize a low-pass filter to remove high-frequency noise. By using low-pass filtering, we can better preserve the meaningful echolocation frequencies produced by bats. By manipulating the cutoff frequency through low-pass filtering, we can control the highest frequency that will be allowed to pass through the filter. An effective pipeline we may use contains the following steps: 1. load data, 2. initiate spectral analysis, 3. apply low-pass filtering, 4. compare results across species, 5. conduct statistical testing (i.e., compare frequencies across species to see if the results are significant), and 6. check for statistically significant outliers.
